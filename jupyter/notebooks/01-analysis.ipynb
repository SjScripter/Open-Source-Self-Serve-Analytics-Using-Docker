# Sales Analytics with Python & Pandas

In this notebook, we will connect to our PostgreSQL database, query the data using SQLAlchemy, and analyze it using Pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
import os

# Read connection secrets from environment variables (passed by Docker Compose)
db_user = os.environ.get('POSTGRES_USER', 'analytics_user')
db_password = os.environ.get('POSTGRES_PASSWORD', 'default_password')
db_name = os.environ.get('POSTGRES_DB', 'analytics_db')

# The hostname is 'postgres' because of Docker's internal networking
engine = create_engine(f'postgresql://{db_user}:{db_password}@postgres:5432/{db_name}')
```

### Load the Data
Let's write a SQL query to get the total revenue per product category.

```python
query = """
SELECT 
    p.category, 
    SUM(oi.quantity * oi.unit_price) as total_revenue
FROM products p
JOIN order_items oi ON p.product_id = oi.product_id
GROUP BY p.category
"""

df = pd.read_sql(query, engine)
df
```

### Visualize the Data

```python
plt.figure(figsize=(8, 5))
plt.bar(df['category'], df['total_revenue'], color='skyblue')
plt.title('Total Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Revenue ($)')
plt.show()
```
